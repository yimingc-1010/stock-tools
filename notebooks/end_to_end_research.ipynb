# End-to-end research workflow (ST-007)

This notebook validates: fetch -> store -> load -> plot for one TWSE symbol.

```python
from stock_tools.core.config import Settings
from stock_tools.data.providers.finmind import FinMindPriceProvider
from stock_tools.data.datasets.roundtrip import build_empty_adjustment_factor_frame, roundtrip_prices_and_adjustments
from stock_tools.data.storage.parquet import ParquetStore
```

```python
symbol = '2330'
start_date = '2024-01-02'
end_date = '2024-01-10'

settings = Settings()
provider = FinMindPriceProvider(api_token=settings.finmind_api_token)
store = ParquetStore(root_dir=settings.data_dir)

daily_prices = provider.fetch_daily_prices(symbol=symbol, start_date=start_date, end_date=end_date)
adjustments = build_empty_adjustment_factor_frame()
result = roundtrip_prices_and_adjustments(
    store=store,
    symbol=symbol,
    daily_prices=daily_prices,
    adjustment_factors=adjustments,
)
result
```

```python
result.daily_prices[['date', 'close']].set_index('date').plot(title=f'{symbol} close price')
```
